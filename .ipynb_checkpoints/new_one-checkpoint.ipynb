```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import numpy as np
import os
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import pandas as pd

def create_enhanced_model(input_shape=(224, 224, 3)):
    """
    Create an enhanced CNN model with improved architecture
    """
    model = Sequential()
    
    # First Convolutional Block
    model.add(Conv2D(32, (3, 3), activation='relu', input_shape=input_shape))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Second Convolutional Block
    model.add(Conv2D(64, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Third Convolutional Block
    model.add(Conv2D(128, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fourth Convolutional Block
    model.add(Conv2D(256, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fifth Convolutional Block
    model.add(Conv2D(512, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.4))
    
    # Flatten and Dense Layers
    model.add(Flatten())
    model.add(Dense(512, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    model.add(Dense(256, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    # Output Layer
    model.add(Dense(4, activation='softmax'))  # 4 classes: glioma, meningioma, notumor, pituitary
    
    # Compile the model
    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    
    return model

def train_enhanced_model():
    """
    Train the enhanced model with better data augmentation and callbacks
    """
    # Create directories if they don't exist
    os.makedirs('models', exist_ok=True)
    os.makedirs('training_history', exist_ok=True)
    
    # Enhanced Data generators with more augmentation
    train_datagen = ImageDataGenerator(
        rescale=1./255,
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.3,
        horizontal_flip=True,
        fill_mode='nearest',
        validation_split=0.2
    )
    
    test_datagen = ImageDataGenerator(rescale=1./255)
    
    # Create separate generators for training and validation
    train_generator = train_datagen.flow_from_directory(
        '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training',
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='training',
        shuffle=True
    )
    
    validation_generator = train_datagen.flow_from_directory(
        '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training',
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='validation',
        shuffle=False
    )
    
    # Create test generator from Testing directory
    test_generator = test_datagen.flow_from_directory(
        '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Testing',
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        shuffle=False
    )
    
    # Check if generators found images
    print(f"Training images found: {train_generator.samples}")
    print(f"Validation images found: {validation_generator.samples}")
    print(f"Test images found: {test_generator.samples}")
    
    # If validation or test generators found 0 images, handle the error
    if validation_generator.samples == 0 or test_generator.samples == 0:
        print("Error: No images found in validation or test sets!")
        print("Please check your dataset structure:")
        print("- Training images should be in 'dataset/Training/class_name/*.jpg'")
        print("- Testing images should be in 'dataset/Testing/class_name/*.jpg'")
        print("- Make sure each class has its own subdirectory")
        return None, None, 0
    
    # Create and train model
    model = create_enhanced_model()
    
    # Display model summary
    model.summary()
    
    # Callbacks
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.00001)
    
    # Train the model
    history = model.fit(
        train_generator,
        epochs=30,
        validation_data=validation_generator,
        callbacks=[early_stop, reduce_lr],
        verbose=1
    )
    
    # Evaluate on test set
    test_loss, test_accuracy = model.evaluate(test_generator)
    print(f"Test Accuracy: {test_accuracy:.4f}")
    
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator, verbose=1)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    class_labels = list(test_generator.class_indices.keys())
    
    # Generate and save confusion matrix
    cm = confusion_matrix(true_classes, predicted_classes)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('training_history/confusion_matrix.png')
    plt.close()
    
    # Generate and save classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels)
    with open('training_history/classification_report.txt', 'w') as f:
        f.write(report)
    
    # Plot training history
    plt.figure(figsize=(12, 5))
    
    # Plot accuracy
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    
    # Plot loss
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    
    plt.tight_layout()
    plt.savefig('training_history/training_history.png')
    plt.close()
    
    # Save the model
    model.save('models/brain_tumor_model.h5')
    
    return model, history, test_accuracy

if __name__ == "__main__":
    train_enhanced_model()
```

```python
def evaluate_model_performance(model_path, test_data_dir):
    """
    Evaluate the trained model and generate detailed performance metrics
    """
    # Load the trained model
    model = tf.keras.models.load_model(model_path)
    
    # Create test data generator
    test_datagen = ImageDataGenerator(rescale=1./255)
    test_generator = test_datagen.flow_from_directory(
        test_data_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        shuffle=False
    )
    
    # Evaluate model
    test_loss, test_accuracy = model.evaluate(test_generator)
    print(f"Test Accuracy: {test_accuracy:.4f}")
    print(f"Test Loss: {test_loss:.4f}")
    
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    class_labels = list(test_generator.class_indices.keys())
    
    # Detailed classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels)
    print("Classification Report:")
    print(report)
    
    # Confusion matrix
    cm = confusion_matrix(true_classes, predicted_classes)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('evaluation/confusion_matrix_detailed.png')
    plt.close()
    
    # ROC Curve and AUC (for multi-class)
    from sklearn.preprocessing import label_binarize
    from sklearn.metrics import roc_curve, auc, roc_auc_score
    
    # Binarize the output for ROC curve
    y_test_bin = label_binarize(true_classes, classes=[0, 1, 2, 3])
    
    # Calculate ROC curve and ROC area for each class
    fpr = {}
    tpr = {}
    roc_auc = {}
    for i in range(4):
        fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], predictions[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = ['blue', 'red', 'green', 'orange']
    for i, color in zip(range(4), colors):
        plt.plot(fpr[i], tpr[i], color=color, 
                 label='ROC curve of class {0} (area = {1:0.2f})'.format(class_labels[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc="lower right")
    plt.savefig('evaluation/roc_curve.png')
    plt.close()
    
    return test_accuracy, report
```

```python
def predict_tumor(image_path, model_path):
    """
    Predict brain tumor type for a single image
    """
    # Load the trained model
    model = tf.keras.models.load_model(model_path)
    
    # Class labels
    class_labels = ['glioma', 'meningioma', 'notumor', 'pituitary']
    
    # Load and preprocess the image
    img = tf.keras.preprocessing.image.load_img(image_path, target_size=(224, 224))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)  # Create batch dimension
    img_array /= 255.0  # Rescale
    
    # Make prediction
    predictions = model.predict(img_array)
    predicted_class = np.argmax(predictions, axis=1)[0]
    confidence = np.max(predictions)
    
    # Display results
    plt.imshow(img)
    plt.title(f'Prediction: {class_labels[predicted_class]}\nConfidence: {confidence:.2f}')
    plt.axis('off')
    plt.show()
    
    return class_labels[predicted_class], confidence

# Example usage:
# predict_tumor('D:\DEEPAK\Me_projects\Major_project\Brain tumor detection\Final_project\data_set\Training\pituitary\Tr-pi_0058.jpg', 'models/brain_tumor_model.h5')
# predict_tumor('D:\DEEPAK\Me_projects\Major_project\Brain tumor detection\Final_project\data_set\Training\glioma\Tr-gl_0716.jpg', 'models/brain_tumor_model.h5')
predict_tumor('D:\DEEPAK\Me_projects\Major_project\Brain tumor detection\Final_project\data_set\Training\meningioma\Tr-me_0011.jpg', 'models/brain_tumor_model.h5')
```

```python
import streamlit as st
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
from PIL import Image
import os

# Set page config
st.set_page_config(
    page_title="Brain Tumor Detection",
    page_icon="🧠",
    layout="wide"
)

# Load model
@st.cache_resource
def load_trained_model(model_path):
    return load_model(model_path)

# Preprocess image
def preprocess_image(image):
    img = image.resize((224, 224))
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array /= 255.0
    return img_array

# Main app
def main():
    st.title("🧠 Brain Tumor Detection")
    st.write("Upload an MRI scan to detect the presence and type of brain tumor")
    
    # File uploader
    uploaded_file = st.file_uploader("Choose an MRI image...", type=["jpg", "jpeg", "png"])
    
    if uploaded_file is not None:
        # Display the uploaded image
        image = Image.open(uploaded_file)
        st.image(image, caption="Uploaded MRI Scan", use_column_width=True)
        
        # Preprocess and predict
        if st.button("Analyze Image"):
            with st.spinner("Analyzing..."):
                # Load model
                model = load_trained_model('models/brain_tumor_model.h5')
                
                # Preprocess image
                processed_image = preprocess_image(image)
                
                # Make prediction
                predictions = model.predict(processed_image)
                predicted_class = np.argmax(predictions, axis=1)[0]
                confidence = np.max(predictions)
                
                # Class labels
                class_labels = ['Glioma', 'Meningioma', 'No Tumor', 'Pituitary']
                
                # Display results
                st.success("Analysis Complete!")
                st.subheader(f"Prediction: {class_labels[predicted_class]}")
                st.subheader(f"Confidence: {confidence*100:.2f}%")
                
                # Show probability distribution
                st.write("Probability Distribution:")
                for i, label in enumerate(class_labels):
                    st.write(f"{label}: {predictions[0][i]*100:.2f}%")

if __name__ == "__main__":
    main()
```

```python
def convert_to_tflite(model_path, output_path):
    """Convert the model to TensorFlow Lite format"""
    converter = tf.lite.TFLiteConverter.from_keras_model(tf.keras.models.load_model(model_path))
    tflite_model = converter.convert()
    
    with open(output_path, 'wb') as f:
        f.write(tflite_model)
    
    print(f"Model converted to {output_path}")

# Convert model
convert_to_tflite('models/brain_tumor_model.h5', 'models/brain_tumor_model.tflite')
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.utils import plot_model
import numpy as np
import os
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, roc_auc_score
from sklearn.preprocessing import label_binarize
import seaborn as sns
import pandas as pd
import cv2
from PIL import Image
import random

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

def create_enhanced_model(input_shape=(224, 224, 3)):
    """
    Create an enhanced CNN model with improved architecture
    """
    model = Sequential()
    
    # First Convolutional Block
    model.add(Conv2D(32, (3, 3), activation='relu', input_shape=input_shape, padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Second Convolutional Block
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Third Convolutional Block
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fourth Convolutional Block
    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fifth Convolutional Block
    model.add(Conv2D(512, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(512, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.4))
    
    # Flatten and Dense Layers
    model.add(Flatten())
    model.add(Dense(512, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    model.add(Dense(256, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    # Output Layer
    model.add(Dense(4, activation='softmax'))  # 4 classes: glioma, meningioma, notumor, pituitary
    
    # Compile the model
    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy', 
                           tf.keras.metrics.Precision(name='precision'),
                           tf.keras.metrics.Recall(name='recall')])
    
    return model

def is_brain_image(image_path, threshold=0.3):
    """
    Determine if an image is likely a brain MRI/CT scan
    Returns True if brain image, False otherwise
    """
    try:
        # Load image in grayscale
        img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        
        if img is None:
            return False
        
        # Resize for consistency
        img = cv2.resize(img, (224, 224))
        
        # Calculate image properties that distinguish brain scans
        # 1. Edge density (brain scans typically have many edges)
        edges = cv2.Canny(img, 50, 150)
        edge_density = np.sum(edges > 0) / (img.shape[0] * img.shape[1])
        
        # 2. Histogram analysis (brain scans have characteristic intensity distributions)
        hist = cv2.calcHist([img], [0], None, [256], [0, 256])
        hist = hist / np.sum(hist)  # Normalize
        
        # Brain scans typically have a bimodal distribution
        # Find peaks in the histogram
        peaks = []
        for i in range(1, len(hist)-1):
            if hist[i] > hist[i-1] and hist[i] > hist[i+1]:
                peaks.append(i)
        
        # 3. Check if the image has the characteristic circular shape of brain scans
        # Create a mask for the circular region
        center = (img.shape[1] // 2, img.shape[0] // 2)
        radius = min(center[0], center[1])
        y, x = np.ogrid[:img.shape[0], :img.shape[1]]
        mask = (x - center[0])**2 + (y - center[1])**2 <= radius**2
        
        # Calculate intensity difference between center and periphery
        center_intensity = np.mean(img[mask])
        periphery_intensity = np.mean(img[~mask])
        intensity_diff = abs(center_intensity - periphery_intensity) / 255
        
        # Decision logic
        is_brain = (
            edge_density > 0.1 and  # Reasonable edge density
            len(peaks) >= 2 and     # At least two peaks in histogram
            intensity_diff > 0.2    # Significant center-periphery difference
        )
        
        return is_brain
    except Exception as e:
        print(f"Error in brain image detection: {e}")
        return False

def visualize_data_samples(data_dir, class_names, num_samples=5):
    """
    Visualize sample images from each class
    """
    plt.figure(figsize=(15, 10))
    
    for i, class_name in enumerate(class_names):
        class_dir = os.path.join(data_dir, class_name)
        if not os.path.exists(class_dir):
            continue
            
        image_files = os.listdir(class_dir)[:num_samples]
        
        for j, image_file in enumerate(image_files):
            img_path = os.path.join(class_dir, image_file)
            img = Image.open(img_path)
            
            plt.subplot(len(class_names), num_samples, i * num_samples + j + 1)
            plt.imshow(img, cmap='gray' if len(np.array(img).shape) == 2 else None)
            plt.title(f'{class_name}' if j == num_samples//2 else '')
            plt.axis('off')
    
    plt.tight_layout()
    plt.savefig('training_history/data_samples.png', dpi=300, bbox_inches='tight')
    plt.close()

def train_enhanced_model():
    """
    Train the enhanced model with better data augmentation and callbacks
    """
    # Create directories if they don't exist
    os.makedirs('models', exist_ok=True)
    os.makedirs('training_history', exist_ok=True)
    os.makedirs('evaluation', exist_ok=True)
    
    # Define dataset paths
    train_dir = '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training'
    test_dir = '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Testing'
    
    # Check if directories exist
    if not os.path.exists(train_dir) or not os.path.exists(test_dir):
        print(f"Error: Dataset directories not found!")
        print(f"Training directory: {train_dir}")
        print(f"Testing directory: {test_dir}")
        return None, None, 0
    
    # Get class names
    class_names = sorted([d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))])
    print(f"Found classes: {class_names}")
    
    # Visualize sample data
    visualize_data_samples(train_dir, class_names)
    
    # Enhanced Data generators with more augmentation
    train_datagen = ImageDataGenerator(
        rescale=1./255,
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.3,
        horizontal_flip=True,
        vertical_flip=True,
        brightness_range=[0.8, 1.2],
        fill_mode='nearest',
        validation_split=0.2
    )
    
    test_datagen = ImageDataGenerator(rescale=1./255)
    
    # Create separate generators for training and validation
    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='training',
        shuffle=True,
        classes=class_names
    )
    
    validation_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='validation',
        shuffle=False,
        classes=class_names
    )
    
    # Create test generator from Testing directory
    test_generator = test_datagen.flow_from_directory(
        test_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        shuffle=False,
        classes=class_names
    )
    
    # Save the class indices for prediction
    class_indices = train_generator.class_indices
    with open('models/class_indices.txt', 'w') as f:
        f.write(str(class_indices))
    
    # Check if generators found images
    print(f"Training images found: {train_generator.samples}")
    print(f"Validation images found: {validation_generator.samples}")
    print(f"Test images found: {test_generator.samples}")
    
    # If validation or test generators found 0 images, handle the error
    if validation_generator.samples == 0 or test_generator.samples == 0:
        print("Error: No images found in validation or test sets!")
        print("Please check your dataset structure:")
        print("- Training images should be in 'dataset/Training/class_name/*.jpg'")
        print("- Testing images should be in 'dataset/Testing/class_name/*.jpg'")
        print("- Make sure each class has its own subdirectory")
        return None, None, 0
    
    # Create and train model
    model = create_enhanced_model()
    
    # Display model summary
    model.summary()
    
    # Plot model architecture
    plot_model(model, to_file='training_history/model_architecture.png', show_shapes=True, show_layer_names=True)
    
    # Callbacks
    early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=1)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.00001, verbose=1)
    checkpoint = ModelCheckpoint('models/best_model.h5', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1)
    
    # Train the model
    history = model.fit(
        train_generator,
        epochs=50,
        validation_data=validation_generator,
        callbacks=[early_stop, reduce_lr, checkpoint],
        verbose=1
    )
    
    # Load the best model
    model = tf.keras.models.load_model('models/best_model.h5')
    
    # Evaluate on test set
    test_results = model.evaluate(test_generator)
    test_loss, test_accuracy, test_precision, test_recall = test_results
    print(f"Test Accuracy: {test_accuracy:.4f}")
    print(f"Test Precision: {test_precision:.4f}")
    print(f"Test Recall: {test_recall:.4f}")
    
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator, verbose=1)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    class_labels = list(test_generator.class_indices.keys())
    
    # Generate and save confusion matrix
    cm = confusion_matrix(true_classes, predicted_classes)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('training_history/confusion_matrix.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Generate and save classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels)
    with open('training_history/classification_report.txt', 'w') as f:
        f.write(report)
    print("Classification Report:")
    print(report)
    
    # Plot training history
    plt.figure(figsize=(15, 5))
    
    # Plot accuracy
    plt.subplot(1, 3, 1)
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    
    # Plot loss
    plt.subplot(1, 3, 2)
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot precision and recall
    plt.subplot(1, 3, 3)
    plt.plot(history.history['precision'], label='Training Precision')
    plt.plot(history.history['val_precision'], label='Validation Precision')
    plt.plot(history.history['recall'], label='Training Recall')
    plt.plot(history.history['val_recall'], label='Validation Recall')
    plt.title('Precision and Recall')
    plt.xlabel('Epoch')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.savefig('training_history/training_history.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Save the final model
    model.save('models/brain_tumor_model.h5')
    
    # Evaluate model performance with detailed metrics
    evaluate_model_performance(model, test_generator, class_labels)
    
    return model, history, test_accuracy

def evaluate_model_performance(model, test_generator, class_labels):
    """
    Evaluate the trained model and generate detailed performance metrics
    """
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator, verbose=1)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    
    # Detailed classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels, output_dict=True)
    
    # Convert to DataFrame for better visualization
    report_df = pd.DataFrame(report).transpose()
    report_df.to_csv('evaluation/detailed_classification_report.csv')
    
    # Plot class-wise metrics
    metrics = ['precision', 'recall', 'f1-score']
    class_metrics = report_df.iloc[:-3, :3]
    
    plt.figure(figsize=(12, 6))
    class_metrics.plot(kind='bar', rot=45)
    plt.title('Class-wise Performance Metrics')
    plt.ylabel('Score')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('evaluation/class_wise_metrics.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # ROC Curve and AUC (for multi-class)
    y_test_bin = label_binarize(true_classes, classes=[0, 1, 2, 3])
    
    # Calculate ROC curve and ROC area for each class
    fpr = {}
    tpr = {}
    roc_auc = {}
    for i in range(len(class_labels)):
        fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], predictions[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = ['blue', 'red', 'green', 'orange']
    for i, color in zip(range(len(class_labels)), colors):
        plt.plot(fpr[i], tpr[i], color=color, 
                 label='ROC curve of {0} (area = {1:0.2f})'.format(class_labels[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.savefig('evaluation/roc_curve.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Calculate and plot macro-average ROC curve
    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(len(class_labels))]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(len(class_labels)):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= len(class_labels)
    
    fpr["macro"] = all_fpr
    tpr["macro"] = mean_tpr
    roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])
    
    plt.figure(figsize=(10, 8))
    for i, color in zip(range(len(class_labels)), colors):
        plt.plot(fpr[i], tpr[i], color=color, alpha=0.6,
                 label='ROC of {0} (area = {1:0.2f})'.format(class_labels[i], roc_auc[i]))
    
    plt.plot(fpr["macro"], tpr["macro"],
             label='Macro-average ROC (area = {0:0.2f})'.format(roc_auc["macro"]),
             color='navy', linestyle=':', linewidth=4)
    
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Multi-class ROC Curves with Macro Average')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.savefig('evaluation/macro_roc_curve.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    return report_df

def predict_tumor(image_path, model_path, confidence_threshold=0.7):
    """
    Predict brain tumor type for a single image with validation
    """
    # First check if the image is a brain scan
    if not is_brain_image(image_path):
        img = Image.open(image_path)
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.title('Invalid Image: This does not appear to be a brain MRI/CT scan')
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        return "invalid_image", 0.0
    
    # Load the trained model
    model = tf.keras.models.load_model(model_path)
    
    # Load class indices
    with open('models/class_indices.txt', 'r') as f:
        class_indices = eval(f.read())
    
    # Create class labels from indices (sort by index value)
    class_labels = [k for k, v in sorted(class_indices.items(), key=lambda item: item[1])]
    
    # Load and preprocess the image
    img = tf.keras.preprocessing.image.load_img(image_path, target_size=(224, 224))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)  # Create batch dimension
    img_array /= 255.0  # Rescale
    
    # Make prediction
    predictions = model.predict(img_array)
    predicted_class_idx = np.argmax(predictions, axis=1)[0]
    confidence = np.max(predictions)
    
    # Check if confidence is above threshold
    if confidence < confidence_threshold:
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.title(f'Uncertain Prediction: Low confidence ({confidence:.2%})\nPlease use a clearer brain MRI image.')
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        return "low_confidence", confidence
    
    # Get the predicted class label
    predicted_class = class_labels[predicted_class_idx]
    
    # Create a visualization with prediction probabilities
    plt.figure(figsize=(15, 5))
    
    # Show the image
    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title(f'Prediction: {predicted_class}\nConfidence: {confidence:.2%}')
    plt.axis('off')
    
    # Show the prediction probabilities as a bar chart
    plt.subplot(1, 2, 2)
    colors = ['red' if i == predicted_class_idx else 'blue' for i in range(len(class_labels))]
    bars = plt.bar(range(len(class_labels)), predictions[0], color=colors, alpha=0.7)
    plt.xlabel('Classes')
    plt.ylabel('Probability')
    plt.title('Prediction Probabilities')
    plt.xticks(range(len(class_labels)), class_labels, rotation=45)
    
    # Add value labels on top of each bar
    for i, bar in enumerate(bars):
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{height:.3f}', ha='center', va='bottom')
    
    plt.ylim(0, 1.1)
    plt.tight_layout()
    plt.show()
    
    # Print all class probabilities
    print("\nPrediction Probabilities:")
    for i, label in enumerate(class_labels):
        print(f"{label}: {predictions[0][i]:.4f} ({predictions[0][i]*100:.2f}%)")
    
    return predicted_class, confidence

def main():
    """
    Main function to run the complete brain tumor detection pipeline
    """
    print("Starting Brain Tumor Detection Pipeline...")
    
    # Step 1: Train the model
    print("\n1. Training the model...")
    model, history, test_accuracy = train_enhanced_model()
    
    if model is None:
        print("Training failed. Exiting.")
        return
    
    print(f"\nModel trained successfully with test accuracy: {test_accuracy:.4f}")
    
    # Step 2: Test with sample images
    print("\n2. Testing with sample images...")
    
    # Define sample images to test
    sample_images = [
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/pituitary/Tr-pi_0058.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/glioma/Tr-gl_0716.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/meningioma/Tr-me_0011.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/notumor/Tr-no_0065.jpg'
    ]
    
    for img_path in sample_images:
        if os.path.exists(img_path):
            print(f"\nTesting image: {os.path.basename(img_path)}")
            predicted_class, confidence = predict_tumor(img_path, 'models/brain_tumor_model.h5')
            print(f"Final Prediction: {predicted_class} with {confidence:.2%} confidence")
        else:
            print(f"Image not found: {img_path}")
    
    # Step 3: Test with a non-brain image (if available)
    print("\n3. Testing with a non-brain image...")
    # You can add a path to a non-brain image here to test the validation
    # non_brain_image = 'path/to/non_brain_image.jpg'
    # if os.path.exists(non_brain_image):
    #     predicted_class, confidence = predict_tumor(non_brain_image, 'models/brain_tumor_model.h5')
    #     print(f"Non-brain image result: {predicted_class}")
    
    print("\nPipeline completed successfully!")
    print("Check the 'training_history' and 'evaluation' folders for detailed results and visualizations.")

if __name__ == "__main__":
    main()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.utils import plot_model
import numpy as np
import os
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, roc_auc_score
from sklearn.preprocessing import label_binarize
import seaborn as sns
import pandas as pd
import cv2
from PIL import Image
import random

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

def create_enhanced_model(input_shape=(224, 224, 3)):
    """
    Create an enhanced CNN model with improved architecture
    """
    model = Sequential()
    
    # First Convolutional Block
    model.add(Conv2D(32, (3, 3), activation='relu', input_shape=input_shape, padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Second Convolutional Block
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.2))
    
    # Third Convolutional Block
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fourth Convolutional Block
    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.3))
    
    # Fifth Convolutional Block
    model.add(Conv2D(512, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(512, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(2, 2))
    model.add(Dropout(0.4))
    
    # Flatten and Dense Layers
    model.add(Flatten())
    model.add(Dense(512, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    model.add(Dense(256, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    
    # Output Layer
    model.add(Dense(4, activation='softmax'))  # 4 classes: glioma, meningioma, notumor, pituitary
    
    # Compile the model
    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy', 
                           tf.keras.metrics.Precision(name='precision'),
                           tf.keras.metrics.Recall(name='recall')])
    
    return model

def is_brain_image(image_path, threshold=0.3):
    """
    Determine if an image is likely a brain MRI/CT scan
    Returns True if brain image, False otherwise
    """
    try:
        # Load image in grayscale
        img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        
        if img is None:
            return False
        
        # Resize for consistency
        img = cv2.resize(img, (224, 224))
        
        # Calculate image properties that distinguish brain scans
        # 1. Edge density (brain scans typically have many edges)
        edges = cv2.Canny(img, 50, 150)
        edge_density = np.sum(edges > 0) / (img.shape[0] * img.shape[1])
        
        # 2. Histogram analysis (brain scans have characteristic intensity distributions)
        hist = cv2.calcHist([img], [0], None, [256], [0, 256])
        hist = hist / np.sum(hist)  # Normalize
        
        # Brain scans typically have a bimodal distribution
        # Find peaks in the histogram
        peaks = []
        for i in range(1, len(hist)-1):
            if hist[i] > hist[i-1] and hist[i] > hist[i+1]:
                peaks.append(i)
        
        # 3. Check if the image has the characteristic circular shape of brain scans
        # Create a mask for the circular region
        center = (img.shape[1] // 2, img.shape[0] // 2)
        radius = min(center[0], center[1])
        y, x = np.ogrid[:img.shape[0], :img.shape[1]]
        mask = (x - center[0])**2 + (y - center[1])**2 <= radius**2
        
        # Calculate intensity difference between center and periphery
        center_intensity = np.mean(img[mask])
        periphery_intensity = np.mean(img[~mask])
        intensity_diff = abs(center_intensity - periphery_intensity) / 255
        
        # Decision logic
        is_brain = (
            edge_density > 0.1 and  # Reasonable edge density
            len(peaks) >= 2 and     # At least two peaks in histogram
            intensity_diff > 0.2    # Significant center-periphery difference
        )
        
        return is_brain
    except Exception as e:
        print(f"Error in brain image detection: {e}")
        return False

def visualize_data_samples(data_dir, class_names, num_samples=5):
    """
    Visualize sample images from each class
    """
    plt.figure(figsize=(15, 10))
    
    for i, class_name in enumerate(class_names):
        class_dir = os.path.join(data_dir, class_name)
        if not os.path.exists(class_dir):
            continue
            
        image_files = os.listdir(class_dir)[:num_samples]
        
        for j, image_file in enumerate(image_files):
            img_path = os.path.join(class_dir, image_file)
            img = Image.open(img_path)
            
            plt.subplot(len(class_names), num_samples, i * num_samples + j + 1)
            plt.imshow(img, cmap='gray' if len(np.array(img).shape) == 2 else None)
            plt.title(f'{class_name}' if j == num_samples//2 else '')
            plt.axis('off')
    
    plt.tight_layout()
    plt.savefig('training_history/data_samples.png', dpi=300, bbox_inches='tight')
    plt.close()

def train_enhanced_model():
    """
    Train the enhanced model with better data augmentation and callbacks
    """
    # Create directories if they don't exist
    os.makedirs('models', exist_ok=True)
    os.makedirs('training_history', exist_ok=True)
    os.makedirs('evaluation', exist_ok=True)
    
    # Define dataset paths
    train_dir = '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training'
    test_dir = '/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Testing'
    
    # Check if directories exist
    if not os.path.exists(train_dir) or not os.path.exists(test_dir):
        print(f"Error: Dataset directories not found!")
        print(f"Training directory: {train_dir}")
        print(f"Testing directory: {test_dir}")
        return None, None, 0
    
    # Get class names
    class_names = sorted([d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))])
    print(f"Found classes: {class_names}")
    
    # Visualize sample data
    visualize_data_samples(train_dir, class_names)
    
    # Enhanced Data generators with more augmentation
    train_datagen = ImageDataGenerator(
        rescale=1./255,
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.3,
        horizontal_flip=True,
        vertical_flip=True,
        brightness_range=[0.8, 1.2],
        fill_mode='nearest',
        validation_split=0.2
    )
    
    test_datagen = ImageDataGenerator(rescale=1./255)
    
    # Create separate generators for training and validation
    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='training',
        shuffle=True,
        classes=class_names
    )
    
    validation_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='validation',
        shuffle=False,
        classes=class_names
    )
    
    # Create test generator from Testing directory
    test_generator = test_datagen.flow_from_directory(
        test_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        shuffle=False,
        classes=class_names
    )
    
    # Save the class indices for prediction
    class_indices = train_generator.class_indices
    with open('models/class_indices.txt', 'w') as f:
        f.write(str(class_indices))
    
    # Check if generators found images
    print(f"Training images found: {train_generator.samples}")
    print(f"Validation images found: {validation_generator.samples}")
    print(f"Test images found: {test_generator.samples}")
    
    # If validation or test generators found 0 images, handle the error
    if validation_generator.samples == 0 or test_generator.samples == 0:
        print("Error: No images found in validation or test sets!")
        print("Please check your dataset structure:")
        print("- Training images should be in 'dataset/Training/class_name/*.jpg'")
        print("- Testing images should be in 'dataset/Testing/class_name/*.jpg'")
        print("- Make sure each class has its own subdirectory")
        return None, None, 0
    
    # Create and train model
    model = create_enhanced_model()
    
    # Display model summary
    model.summary()
    
    # Plot model architecture
    try:
        plot_model(model, to_file='training_history/model_architecture.png', show_shapes=True, show_layer_names=True)
    except:
        print("Note: pydot not installed. Skipping model architecture plot.")
    
    # Callbacks - FIXED: Changed file extension from .h5 to .keras
    early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=1)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.00001, verbose=1)
    checkpoint = ModelCheckpoint('models/best_model.keras', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1)
    
    # Train the model
    history = model.fit(
        train_generator,
        epochs=50,
        validation_data=validation_generator,
        callbacks=[early_stop, reduce_lr, checkpoint],
        verbose=1
    )
    
    # Load the best model
    model = tf.keras.models.load_model('models/best_model.keras')
    
    # Evaluate on test set
    test_results = model.evaluate(test_generator)
    test_loss, test_accuracy, test_precision, test_recall = test_results
    print(f"Test Accuracy: {test_accuracy:.4f}")
    print(f"Test Precision: {test_precision:.4f}")
    print(f"Test Recall: {test_recall:.4f}")
    
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator, verbose=1)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    class_labels = list(test_generator.class_indices.keys())
    
    # Generate and save confusion matrix
    cm = confusion_matrix(true_classes, predicted_classes)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('training_history/confusion_matrix.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Generate and save classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels)
    with open('training_history/classification_report.txt', 'w') as f:
        f.write(report)
    print("Classification Report:")
    print(report)
    
    # Plot training history
    plt.figure(figsize=(15, 5))
    
    # Plot accuracy
    plt.subplot(1, 3, 1)
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    
    # Plot loss
    plt.subplot(1, 3, 2)
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot precision and recall
    plt.subplot(1, 3, 3)
    plt.plot(history.history['precision'], label='Training Precision')
    plt.plot(history.history['val_precision'], label='Validation Precision')
    plt.plot(history.history['recall'], label='Training Recall')
    plt.plot(history.history['val_recall'], label='Validation Recall')
    plt.title('Precision and Recall')
    plt.xlabel('Epoch')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.savefig('training_history/training_history.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Save the final model - FIXED: Changed file extension from .h5 to .keras
    model.save('models/brain_tumor_model.keras')
    
    # Evaluate model performance with detailed metrics
    evaluate_model_performance(model, test_generator, class_labels)
    
    return model, history, test_accuracy

```

```python
def evaluate_model_performance(model, test_generator, class_labels):
    """
    Evaluate the trained model and generate detailed performance metrics
    """
    # Generate predictions
    test_generator.reset()
    predictions = model.predict(test_generator, verbose=1)
    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = test_generator.classes
    
    # Detailed classification report
    report = classification_report(true_classes, predicted_classes, target_names=class_labels, output_dict=True)
    
    # Convert to DataFrame for better visualization
    report_df = pd.DataFrame(report).transpose()
    report_df.to_csv('evaluation/detailed_classification_report.csv')
    
    # Plot class-wise metrics
    metrics = ['precision', 'recall', 'f1-score']
    class_metrics = report_df.iloc[:-3, :3]
    
    plt.figure(figsize=(12, 6))
    class_metrics.plot(kind='bar', rot=45)
    plt.title('Class-wise Performance Metrics')
    plt.ylabel('Score')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('evaluation/class_wise_metrics.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # ROC Curve and AUC (for multi-class)
    y_test_bin = label_binarize(true_classes, classes=[0, 1, 2, 3])
    
    # Calculate ROC curve and ROC area for each class
    fpr = {}
    tpr = {}
    roc_auc = {}
    for i in range(len(class_labels)):
        fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], predictions[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = ['blue', 'red', 'green', 'orange']
    for i, color in zip(range(len(class_labels)), colors):
        plt.plot(fpr[i], tpr[i], color=color, 
                 label='ROC curve of {0} (area = {1:0.2f})'.format(class_labels[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.savefig('evaluation/roc_curve.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # Calculate and plot macro-average ROC curve
    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(len(class_labels))]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(len(class_labels)):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= len(class_labels)
    
    fpr["macro"] = all_fpr
    tpr["macro"] = mean_tpr
    roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])
    
    plt.figure(ffigsize=(10, 8))
    for i, color in zip(range(len(class_labels)), colors):
        plt.plot(fpr[i], tpr[i], color=color, alpha=0.6,
                 label='ROC of {0} (area = {1:0.2f})'.format(class_labels[i], roc_auc[i]))
    
    plt.plot(fpr["macro"], tpr["macro"],
             label='Macro-average ROC (area = {0:0.2f})'.format(roc_auc["macro"]),
             color='navy', linestyle=':', linewidth=4)
    
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Multi-class ROC Curves with Macro Average')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.savefig('evaluation/macro_roc_curve.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    return report_df

def predict_tumor(image_path, model_path, confidence_threshold=0.7):
    """
    Predict brain tumor type for a single image with validation
    """
    # First check if the image is a brain scan
    if not is_brain_image(image_path):
        img = Image.open(image_path)
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.title('Invalid Image: This does not appear to be a brain MRI/CT scan')
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        return "invalid_image", 0.0
    
    # Load the trained model
    model = tf.keras.models.load_model(model_path)
    
    # Load class indices
    with open('models/class_indices.txt', 'r') as f:
        class_indices = eval(f.read())
    
    # Create class labels from indices (sort by index value)
    class_labels = [k for k, v in sorted(class_indices.items(), key=lambda item: item[1])]
    
    # Load and preprocess the image
    img = tf.keras.preprocessing.image.load_img(image_path, target_size=(224, 224))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)  # Create batch dimension
    img_array /= 255.0  # Rescale
    
    # Make prediction
    predictions = model.predict(img_array)
    predicted_class_idx = np.argmax(predictions, axis=1)[0]
    confidence = np.max(predictions)
    
    # Check if confidence is above threshold
    if confidence < confidence_threshold:
        plt.figure(figsize=(8, 8))
        plt.imshow(img)
        plt.title(f'Uncertain Prediction: Low confidence ({confidence:.2%})\nPlease use a clearer brain MRI image.')
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        return "low_confidence", confidence
    
    # Get the predicted class label
    predicted_class = class_labels[predicted_class_idx]
    
    # Create a visualization with prediction probabilities
    plt.figure(figsize=(15, 5))
    
    # Show the image
    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title(f'Prediction: {predicted_class}\nConfidence: {confidence:.2%}')
    plt.axis('off')
    
    # Show the prediction probabilities as a bar chart
    plt.subplot(1, 2, 2)
    colors = ['red' if i == predicted_class_idx else 'blue' for i in range(len(class_labels))]
    bars = plt.bar(range(len(class_labels)), predictions[0], color=colors, alpha=0.7)
    plt.xlabel('Classes')
    plt.ylabel('Probability')
    plt.title('Prediction Probabilities')
    plt.xticks(range(len(class_labels)), class_labels, rotation=45)
    
    # Add value labels on top of each bar
    for i, bar in enumerate(bars):
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{height:.3f}', ha='center', va='bottom')
    
    plt.ylim(0, 1.1)
    plt.tight_layout()
    plt.show()
    
    # Print all class probabilities
    print("\nPrediction Probabilities:")
    for i, label in enumerate(class_labels):
        print(f"{label}: {predictions[0][i]:.4f} ({predictions[0][i]*100:.2f}%)")
    
    return predicted_class, confidence

def main():
    """
    Main function to run the complete brain tumor detection pipeline
    """
    print("Starting Brain Tumor Detection Pipeline...")
    
    # Step 1: Train the model
    print("\n1. Training the model...")
    model, history, test_accuracy = train_enhanced_model()
    
    if model is None:
        print("Training failed. Exiting.")
        return
    
    print(f"\nModel trained successfully with test accuracy: {test_accuracy:.4f}")
    
    # Step 2: Test with sample images
    print("\n2. Testing with sample images...")
    
    # Define sample images to test
    sample_images = [
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/pituitary/Tr-pi_0058.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/glioma/Tr-gl_0716.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/meningioma/Tr-me_0011.jpg',
        'D:/DEEPAK/Me_projects/Major_project/Brain tumor detection/Final_project/data_set/Training/notumor/Tr-no_0065.jpg'
    ]
    
    for img_path in sample_images:
        if os.path.exists(img_path):
            print(f"\nTesting image: {os.path.basename(img_path)}")
            # FIXED: Changed model path from .h5 to .keras
            predicted_class, confidence = predict_tumor(img_path, 'models/brain_tumor_model.keras')
            print(f"Final Prediction: {predicted_class} with {confidence:.2%} confidence")
        else:
            print(f"Image not found: {img_path}")
    
    # Step 3: Test with a non-brain image (if available)
    print("\n3. Testing with a non-brain image...")
    # You can add a path to a non-brain image here to test the validation
    # non_brain_image = 'path/to/non_brain_image.jpg'
    # if os.path.exists(non_brain_image):
    #     predicted_class, confidence = predict_tumor(non_brain_image, 'models/brain_tumor_model.keras')
    #     print(f"Non-brain image result: {predicted_class}")
    
    print("\nPipeline completed successfully!")
    print("Check the 'training_history' and 'evaluation' folders for detailed results and visualizations.")

if __name__ == "__main__":
    main()
```
